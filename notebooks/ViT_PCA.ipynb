```python
# Set the working directory
import os
import numpy as np
import pandas as pd

#work_dir = os.path.dirname(os.path.dirname(os.getcwd())) 
work_dir = os.path.dirname(os.getcwd())

print(f"Current working directory: {work_dir}")
```

```python
import numpy as np
from pathlib import Path
from sklearn.decomposition import PCA
import pickle

PCA_DIM = 256

train_vit = np.load(f"{work_dir}/data/train/vit_features.npy").astype(np.float32)
val_vit = np.load(f"{work_dir}/data/val/vit_features.npy").astype(np.float32)

print("train_vit:", train_vit.shape)
print("val_vit:", val_vit.shape)

# 念のため正規化
train_vit = train_vit / (np.linalg.norm(train_vit, axis=1, keepdims=True) + 1e-6)
val_vit = val_vit / (np.linalg.norm(val_vit, axis=1, keepdims=True) + 1e-6)

pca = PCA(n_components=PCA_DIM, random_state=42)
train_pca = pca.fit_transform(train_vit)
val_pca = pca.transform(val_vit)

train_pca = train_pca.astype(np.float32)
val_pca = val_pca.astype(np.float32)

# PCA後も正規化
train_pca = train_pca / (np.linalg.norm(train_pca, axis=1, keepdims=True) + 1e-6)
val_pca = val_pca / (np.linalg.norm(val_pca, axis=1, keepdims=True) + 1e-6)

np.save(f"{work_dir}/data/train/vit_pca{PCA_DIM}_features.npy", train_pca)
np.save(f"{work_dir}/data/val/vit_pca{PCA_DIM}_features.npy", val_pca)

with open(f"{work_dir}/vit_pca{PCA_DIM}.pkl", "wb") as f:
    pickle.dump(pca, f)

print("saved:")
print(f"{work_dir}/data/train/vit_pca{PCA_DIM}_features.npy", train_pca.shape)
print(f"{work_dir}/data/val/vit_pca{PCA_DIM}_features.npy", val_pca.shape)

print("explained variance ratio sum:", pca.explained_variance_ratio_.sum())
print("train_pca norm mean:", np.linalg.norm(train_pca, axis=1).mean())
print("val_pca norm mean:", np.linalg.norm(val_pca, axis=1).mean())
```
